```python
import pandas as pd
import numpy as np
import nltk
```

```python
from nltk.corpus import stopwords
stop = stopwords.words('english')
```

```python
df = pd.read_csv("/home/devjith/Luminar/Machine Learning/files/DisneylandReviews.csv",encoding='ISO-8859-1')
```

```python
df = df.drop(columns='Review_ID')
```

```python
df
```

```python
df = df.drop(columns=['Year_Month'])
```

```python
df
```

```python
text = df['Review_Text']
```

```python
from nltk.tokenize import TweetTokenizer,word_tokenize
tk = TweetTokenizer()
```

```python
text = text.apply(lambda x: tk.tokenize(x)).apply(lambda x: ' '.join(x))
```

```python
import re
text = text.apply(lambda x: re.sub('[^a-zA-Z0-9]',' ',x))
```

```python
text = text.apply(lambda x: ' '.join([i for i in word_tokenize(x) if len(i)>=3]))
```

```python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer(language='english')

text = text.apply(lambda x : ' '.join([stemmer.stem(i.lower()) for i in word_tokenize(x)]))
```

```python
text = text.apply(lambda x: ' '.join([i for i in word_tokenize(x) if i not in stop]))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vector = TfidfVectorizer()

vect_text = vector.fit_transform(text)
type(vect_text)
```

```python
tfidf_df = pd.DataFrame(vect_text.toarray(), columns=vector.get_feature_names_out())
df = pd.concat([df, tfidf_df], axis=1)
df = df.drop(columns=['Review_Text'])
```

```python
df
```

```python
df_enc = pd.get_dummies(df[['Reviewer_Location','Branch']],drop_first=True,dtype=int)
```

```python
df = df.drop(columns=['Reviewer_Location','Branch'])
df = pd.concat([df,df_enc],axis=1)
```
