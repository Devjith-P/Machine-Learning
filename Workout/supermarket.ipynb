```python
df['Day'] = df['Date'].dt.day
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
df = df.drop(columns='Date')
df
```

```python
df['Time'] = pd.to_datetime(df['Time'])
```

```python
df.dtypes
```

```python
df
```

```python
def time_of_day(hour):
    if 6 <= hour <12:
        return "Morning"
    elif 12<= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"
```

```python
df['Hour'] = df['Time'].dt.hour
```

```python
df['Time_of_Day'] = df['Hour'].apply(time_of_day)
```

```python
df
```

```python
df = df.drop(columns=['cogs','gross margin percentage','gross income','Time'])
```

```python
df
```

```python
df_enc = pd.get_dummies(df[['Branch','City','Customer type','Gender','Product line','Payment','Time_of_Day']],drop_first=True,dtype=int)
```

```python
df = pd.concat([df,df_enc],axis=1)
```

```python
df = df.drop(columns=['Branch','City','Customer type','Gender','Product line','Payment','Time_of_Day'])
```

```python
df
```

```python
df.columns
```

```python
x = df.drop(columns='Rating')
```

```python
y = df['Rating']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=150,n_jobs=-1)

model.fit(x_train,y_train)
```

```python
y_pred  = model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score

mean_absolute_percentage_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```
